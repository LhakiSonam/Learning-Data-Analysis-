## 2. Loading the Dataset

In this section, I will load the Student Performance dataset using Pandas. I will then explore the dataset to understand its structure, columns, and the information it contains.

```python
# Load the dataset

import pandas as pd

from google.colab import files

uploaded = files.upload()

# Read the uploaded CSV file
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

# Display the first five rows
df.head()
```

## 4. Data Cleaning

Data cleaning is the process of identifying and handling missing values and duplicate records in a dataset. In this section, I will check whether the Student Performance dataset contains missing information or repeated records.

```python
# Check for missing values in each column

df.isnull().sum()
```

```python
# Check for duplicate records

df.duplicated().sum()
```

```python
# Remove duplicate records

df = df.drop_duplicates()

print("Duplicate records removed!")
```

```python
# Check the dataset after cleaning

print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
```

## 5. Data Analysis

In this section, I will analyze students' academic performance by calculating the average scores in mathematics, reading, and writing. I will also compare the results to identify which subject has the highest and lowest average scores.

```python
# Calculate average scores for each subject

math_average = df["math score"].mean()
reading_average = df["reading score"].mean()
writing_average = df["writing score"].mean()

print("Average Math Score:", math_average)
print("Average Reading Score:", reading_average)
print("Average Writing Score:", writing_average)
```

```python
# Find the highest and lowest scores

subjects = ["math score", "reading score", "writing score"]

for subject in subjects:
    print(subject)
    print("Highest score:", df[subject].max())
    print("Lowest score:", df[subject].min())
    print()
```

## 6. Data Visualization

In this section, I will use graphs to present the results of the data analysis. Visualizations make it easier to compare students' scores and understand patterns in the dataset.

```python
# Compare average scores in three subjects

average_scores = {
    "Math": df["math score"].mean(),
    "Reading": df["reading score"].mean(),
    "Writing": df["writing score"].mean()
}

plt.figure(figsize=(8, 5))
plt.bar(average_scores.keys(), average_scores.values())

plt.title("Average Scores by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Score")
plt.ylim(0, 100)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df["math score"], bins=10, edgecolor="black")

plt.title("Distribution of Math Scores")
plt.xlabel("Math Scores")
plt.ylabel("Number of Students")

plt.show()
```
